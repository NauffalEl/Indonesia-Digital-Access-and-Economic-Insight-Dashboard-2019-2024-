```python
import pandas as pd
```

```python
file_path = r'H:\Project\Internet access\onedata.xlsx' # Gunakan 'r' untuk path Windows

# Memuat semua sheet
xls = pd.ExcelFile(file_path)
sheet_names = xls.sheet_names
dataframes = {sheet: xls.parse(sheet) for sheet in sheet_names}
```

```python
print(f"### Tipe Data Kolom untuk Sheet: {sheet_names[0]}")
print(dataframes[sheet_names[0]].dtypes)
```

```python
df_sheet1 = dataframes[sheet_names[0]]

print("### Nilai Unik Kolom 'YEAR' :")
print(df_sheet1['YEAR'].unique())
print("\nJumlah unik:", df_sheet1['YEAR'].nunique())
print("-" * 30)

print("### Nilai Unik Kolom 'PROVINCE':")
# Menggunakan .sort_values() untuk melihat secara terurut (jika ingin)
print(df_sheet1['PROVINCE'].unique())
print("\nJumlah unik:", df_sheet1['PROVINCE'].nunique())
print("-" * 30)

print("### Nilai Unik Kolom 'CATEGORY':")
print(df_sheet1['CATEGORY'].unique())
print("\nJumlah unik:", df_sheet1['CATEGORY'].nunique())
print("-" * 30)

print("### Nilai Unik Kolom 'SUB CATEGORY':")
print(df_sheet1['SUB CATEGORY'].unique())
print("\nJumlah unik:", df_sheet1['SUB CATEGORY'].nunique())
print("-" * 30)
```
